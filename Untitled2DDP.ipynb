```python
nilai = int(input("masukkan nilai: "))

if nilai >= 75:
  print("lulus")
else:
  print("tidak lulus")
```

```python
usia = int(input("masukkan usia: "))

if usia >= 18:
  print("dewasa")
elif usia >= 13 and usia <= 17:
  print("remaja")
else:
  print("anak - anak")
```

```python
jumlah = float(input("masukkan jumlah belanja: "))

if jumlah > 100:
  print("total harga setelah diskon: ", jumlah - (jumlah * 10 / 100))
else:
  print("Total harga: ", jumlah)
```

```python
print("silahkan pilih bangun datar")
print("1. Persegi")
print("2. Lingkaran")
print("3. Segitiga")

pilih_bangun_datar = int(input("pilih bangun datar:"))

match pilih_bangun_datar:
   case 1:
      sisi = float(input("masukkan sisi:"))
      luas = sisi * sisi
      print("luas persegi=", luas)

   case 2:
       jari = float(input("masukkan jari jari:"))
       luas = 3.14 * jari ** 2
       print("luas lingkaran=", luas)

   case 3:
       alas = float(input("masukkan alas:"))
       tinggi = float(input("masukkan tinggi:"))
       luas = 0.5 * alas * tinggi
       print("luas segitiga=", luas)

   case _:
       print("kamu salah pilih")
```
